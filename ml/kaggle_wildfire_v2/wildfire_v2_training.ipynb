# Wildfire Risk Model v2 (district x day)

Trains `wildfire_xgb_v2` on the Kaggle dataset `shaikhfardin05/india-district-wildfire-v2`, which is produced by
`ml/pipeline/build_wildfire_dataset.py` (NASA FIRMS VIIRS NOAA-20 SP labels + NASA POWER weather + project FWI code).

The notebook only calls `ml/train/train_wildfire_v2.py` (shipped inside the dataset), so Kaggle and local runs execute
identical code. Split: train 2019-2022, validation 2023, test 2024. Baselines (FWI-only, climatology, persistence)
are computed on the same test rows. Outputs land in `/kaggle/working`.

```python
import glob, os, subprocess, sys

root = os.path.dirname(glob.glob("/kaggle/input/**/wildfire_dataset.parquet", recursive=True)[0])
print("dataset root:", root)
cmd = [
    sys.executable,
    f"{root}/ml/train/train_wildfire_v2.py",
    "--dataset",
    f"{root}/wildfire_dataset.parquet",
    "--climatology",
    f"{root}/climatology.json",
    "--out-dir",
    "/kaggle/working",
]
print(subprocess.run(cmd, check=True, capture_output=True, text=True).stdout)
```
