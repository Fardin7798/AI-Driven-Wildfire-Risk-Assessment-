# PM2.5 sub-index forecaster (district x day, h = 1-3 days)

Trains `aqi_pm25_xgb_v1` on the Kaggle dataset `shaikhfardin05/india-district-aqi-pm25`, produced by
`ml/pipeline/build_aqi_dataset.py` (target: CPCB PM2.5 sub-index of the daily CAMS PM2.5 mean, >= 16 valid hours;
PM10/NO2/SO2/CO/O3 and NASA POWER weather are lagged features only).

The notebook only calls `train_aqi_pm25.py` (shipped inside the dataset, a copy of `ml/train/train_aqi_pm25.py`), so
Kaggle and local runs execute identical code. Split: train <= 2024-10-31, val 2024-11-01..2025-10-31 (tuning),
test 2025-11-01..2026-10-28 (evaluated once). Outputs land in `/kaggle/working`.

```python
import glob
import os
import subprocess
import sys

root = os.path.dirname(glob.glob("/kaggle/input/**/aqi_dataset.parquet", recursive=True)[0])
script = glob.glob("/kaggle/input/**/train_aqi_pm25.py", recursive=True)[0]
print("dataset root:", root, "script:", script)
subprocess.run(
    [
        sys.executable,
        script,
        "--dataset",
        f"{root}/aqi_dataset.parquet",
        "--out-dir",
        "/kaggle/working",
    ],
    check=True,
)
```

```python
import json

ev = json.load(open("/kaggle/working/aqi_pm25_evaluation.json"))
print(json.dumps({k: ev[k] for k in ("rows", "leakage_checks", "val", "test")}, indent=2))
```
